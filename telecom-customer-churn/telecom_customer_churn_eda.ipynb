```python
# Importing Libaries
import pandas as pd
```

```python
# Loading dataset
df = pd.read_csv(r"D:\Data Analyst Stuff\WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
# Quick overview, checking for missing entries, data types etc
print(df.info())
```

```python
# Checking for missing values
print(df.isnull().sum())
```

```python
# Checking for duplicate rows
print(df.duplicated().sum())
```

```python
print(df.dtypes)
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"])
```

```python
# Earlier we had no null entries however upon trying to convert TotalCharges from object to float64 we encountered the error above so inspecting " " entries.
df[df["TotalCharges"] == " "]
```

```python
# Proceeding with conversion. 
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

```python
print(df.isnull().sum())
```

```python
# Inspecting the now null entries now that they aren't just " " to decide whether to fill or drop.
df[df["TotalCharges"].isnull()]
```

```python
# Dropping the 11 null entries. 
df = df.dropna(subset=["TotalCharges"])
```

```python
# Rechecking null entries to make sure there is now none again. 
print(df.isnull().sum())
```

```python
# Checking data types to see if TotalCharges has converted from object to float64
print(df.dtypes)
```

```python
# EDA Section
```

```python
# What's the overall churn rate?
```

```python
df.loc[:, "Churn_Flag"] = df["Churn"].map({"Yes": 1, "No": 0})
```

```python
print(df[["Churn", "Churn_Flag"]].head())
```

```python
churn_rate = df["Churn_Flag"].mean()
print(f"Overall Churn Rate: {churn_rate:.2%}")
```

```python
# Which contract type has the highest churn?
```

```python
contract_churn = (df.groupby("Contract")["Churn_Flag"].mean().sort_values(ascending=False) * 100)

print(contract_churn)
```

```python
# Does tenure affect churn?
```

```python
print(df.groupby("Churn")["tenure"].describe())
```

```python
tenure_avg = df.groupby("Churn")["tenure"].mean()
print(tenure_avg)
```

```python
# Do customers with higher monthly charges churn more?
```

```python
monthly_avg = df.groupby("Churn")["MonthlyCharges"].mean()
print(monthly_avg)
```

```python
# Does Tech Support reduce churn?
```

```python
tech_churn = (df.groupby("TechSupport")["Churn_Flag"].mean() * 100)
print(tech_churn)
```

```python
# Does InternetService affect churn?
```

```python
internet_churn = (df.groupby("InternetService")["Churn_Flag"].mean() * 100)
print(internet_churn)
```

```python
fiber_price_check = df.groupby("InternetService").agg(
    Avg_Monthly_Charges=("MonthlyCharges", "mean"),
    Churn_Rate=("Churn_Flag", lambda x: x.mean() * 100),
    Customer_Count=("Churn_Flag", "count"),
)

print(fiber_price_check.round(2))
```

```python
# Senior vs Non-Senior Churn Rate
```

```python
senior_churn = df.groupby("SeniorCitizen").agg(
    Customer_Count=("Churn_Flag", "count"),
    Avg_Monthly_Charges=("MonthlyCharges", "mean"),
    Churn_Rate=("Churn_Flag", lambda x: x.mean() * 100),
)

print(senior_churn.round(2))
```

```python
# Does Phone Service affect Churn Rate?
```

```python
phone_counts = df["PhoneService"].value_counts()
phone_pcts = df["PhoneService"].value_counts(normalize=True) * 100

phone_summary = pd.DataFrame({"Count": phone_counts, "Percentage (%)": phone_pcts.round(2)})

print(phone_summary)
```

```python
lines_summary = df["MultipleLines"].value_counts(normalize=True) * 100
print(lines_summary.round(2).astype(str) + "%")
```

```python
phone_by_age = pd.crosstab(df["SeniorCitizen"], df["PhoneService"], normalize="index") * 100

phone_by_age.index = ["Non-Senior", "Senior"]
print("--- Phone Service Adoption by Age Group (%) ---")
print(phone_by_age.round(2))
```

```python
# Which contracts opt in for streaming add-ons?
```

```python
# Streaming TV adoption % by Contract Type
tv_by_contract = (pd.crosstab(df["Contract"], df["StreamingTV"], normalize="index")["Yes"]* 100)

# Streaming Movies adoption % by Contract Type
movies_by_contract = (pd.crosstab(df["Contract"], df["StreamingMovies"], normalize="index")["Yes"]* 100)

# Combining results into one table
streaming_contract_summary = pd.DataFrame({"Streaming TV Opt-In %": tv_by_contract, "Streaming Movies Opt-In %": movies_by_contract})

print(streaming_contract_summary.round(2))
```

```python
# Create a helper flag for customers who have BOTH streaming add-ons
df.loc[:, "Both_Streaming"] = (df["StreamingTV"] == "Yes") & (df["StreamingMovies"] == "Yes")

# Aggregate adoption rate of both services by Contract
full_package = df.groupby("Contract")["Both_Streaming"].mean() * 100

print("--- % of Customers with BOTH Streaming TV & Movies ---")
print(full_package.round(2).astype(str) + "%")
```
